```python
# Grafico de dispersão com legendas

import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Carregamneto Data base
base = pd.read_csv('Db/co2.csv')
base.head()
```

```python
# Criando variaveis para cada atributo
x = base.conc
y = base.uptake
```

```python
unicos = list(set(base.Treatment))
unicos
# retorna todos os valores possiveis de Treatment 
```

```python
# Percorre cada tipo de elemento (chilled e nonchilled) e cria um grafico de dispersão
for i in range(len(unicos)):
    indice = base.Treatment == unicos[i]
    plt.scatter(x[indice], y[indice], label = unicos[i])
plt.legend(loc = 'lower right')
```
