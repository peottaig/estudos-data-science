```python
# Boxplot
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Carregamento da base de dados
base = pd.read_csv('Db/trees.csv')
base.head()
```

```python
# Geração do boxplot
# patch_artist = True, preenche; showfliers True, outliers
plt.boxplot(base.Volume, vert = True, showfliers = True, notch = True, patch_artist= True)
```

```python
# Dados por linha
plt.boxplot(base, patch_artist=True)
plt.title('Árvores')
plt.xlabel('Dados')
```

```python
# Geração de 3 boxplots, cada um mostrando informações diferentes
plt.boxplot(base.Volume, vert= True)
plt.boxplot(base.Girth, vert= True)
plt.boxplot(base.Height, vert= True)
plt.title("Árvores")
plt.xlabel("Dados")
```

# Com uso da biblioteca Seaborn

```python
import seaborn as srn
```

```python
# Visualização do boxplot
srn.boxplot(data=base.Volume, orient='v').set_title('Árvores')
```

```python
# Visualizaçãoo de varios boxplots na mesma imagem
srn.boxplot(data = base)
```
