```python
# Histograma
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
base = pd.read_csv('Db/trees.csv')
base.shape
# 31 linhas e 3 colunas
```

```python
#dados
```

```python
base.head()
```

```python
# Crição do histograma, considerando somente o segundo atributo da base de dados e com duas divisões (bina)
# A variavel 'h' armazena as faixas de valores de height
h = np.histogram(base.iloc[:,1], bins = 6)
h
```

```python
# Visualização do histograma com 6 divisões (bins)
plt.hist(base.iloc[:,1], bins = 11)
plt.title('Árvores')
plt.ylabel('Frequencia')
plt.xlabel('Altura')
```

#### Histograma com seaborn

```python
import pandas as pd
import seaborn as srn
import matplotlib.pyplot as plt
```

```python
# Carregamento
base = pd.read_csv('Db/trees.csv')
base.head()
```

```python
# Histograma com 10 divisões (bins) e com gráfico de densidade (kernel density - suaviação)
# stat: count, frequency, probability, percent, density
srn.histplot(base.Volume, color= "blue", kde=True, stat="density", bins=10).set(title='Volume')
```
