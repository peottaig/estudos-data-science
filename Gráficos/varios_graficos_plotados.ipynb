```python
# Divisão de tela (subgraficos)
import pandas as pd
import matplotlib.pyplot as plt
```

```python
base = pd.read_csv('Db/trees.csv')
base.head()
```

##### Graficos separados
* Avaliando a relação dos atributos das arvores

```python
# Girth com volume
plt.scatter(base.Girth, base.Volume)
```

```python
# Girth com height
plt.scatter(base.Girth, base.Height)
```

```python
# Height com volume
plt.scatter(base.Height, base.Volume)
```

```python
# Histograma
plt.hist(base.Volume, bins=7)
```

### Graficos plotados juntos para comparação

```python
# imprimindo juntos
plt.figure(1)
plt.subplot(2,2,1)
plt.scatter(base.Girth, base.Volume)
plt.subplot(2,2,2)
plt.scatter(base.Girth, base.Height)
plt.subplot(2,2,3)
plt.scatter(base.Height, base.Volume)
plt.subplot(2,2,4)
plt.hist(base.Volume, bins=7)


```
