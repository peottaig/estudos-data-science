```python
# Gráfico de barras e setores
import pandas as pd
```

```python
# Carregamento padrão
base = pd.read_csv("Db/insect.csv")
base.shape
```

```python
# Dados
base.head()
```

```python
# Agrupamento dos dados baseado no atributo 'sprey', contando e somando os registros
agrupado = base.groupby(['spray'])['count'].sum()
agrupado
```

```python
# Grafico de barras 
agrupado.plot.bar(color = ['blue', 'yellow', 'red', 'green', 'pink', 'orange'])
# agrupado.plot.bar(color = 'gray')
```

```python
# Grafico de pizza(setores)
agrupado.plot.pie(legend = True)
```
