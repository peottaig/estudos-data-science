```python
# Mudança da seed aleatoria randomica
np.random.seed(2345)
# 150 amostras, de 0 a 1, com reposição, probabilidades equivalentes
amostra = np.random.choice(a = [0, 1], size=150, replace=True,
                           p = [0.7, 0.3])

# Verificação do tamanho da amostra
len(amostra)
```
