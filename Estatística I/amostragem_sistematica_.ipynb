```python
# Amostragem sistematica
import pandas as pd
import numpy as np
from math import ceil
```

```python
# Criação das variaveis para representar a população, a amostra e o valor de k
populacao = 150
amostra = 15
k = ceil(populacao / amostra)
print(k)
```

```python
# Definição do valor randômico para inicializar a amostra, iniciando em 1 até k + 1
r = np.random.randint(low= 1, high= k + 1, size = 1)
print(r)
```

```python
# Criamos um for para somar os proximo valores, baseado no primeiro valor r que foi definido acima
acumulador = r[0]
sorteados = []
for i in range(amostra):
    # print(acumulador)
    sorteados.append(acumulador)
    acumulador += k
print(sorteados)
```

```python
# Carregamento da base de dados e dataframe final com os valores sorteados

dataframe = pd.read_csv("Db/iris.csv")
final_dataframe = dataframe.loc[sorteados]
final_dataframe
```
