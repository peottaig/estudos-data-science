```python
# Teste distribuição normal
from scipy import stats
from scipy.stats import norm, skewnorm
import matplotlib.pyplot as plt
```

```python
# Criação de uma variavel com dados em uma distribuição normal com a função rvs (100 elementos)
dados = norm.rvs(size = 1000)
dados
```

```python
# histograma
plt.hist(dados, bins= 15)
plt.title("Dados")
plt.show()
```

```python
# Geração do grafico para verificar se a distribuição é normal
fig, ax = plt.subplots()
stats.probplot(dados, fit=True, plot=ax)
plt.show()
```

```python
# Execução do teste de Shapiro
# Segundo argumento é o valor de p, não há como rejeitar a hipótese nula. Como observado o p é maior que 0.05
stats.shapiro(dados)
```

```python
# Dados não normais: 
dados_enviesados = skewnorm.rvs(4, size=1000)
```

```python
# histograma
plt.hist(dados_enviesados, bins= 15)
plt.title("Dados")
plt.show()
```

```python
# Geração do grafico para verificar se a distribuição é normal
fig, ax = plt.subplots()
stats.probplot(dados_enviesados, fit=True, plot=ax)
plt.show()
```

```python
stats.shapiro(dados_enviesados)

# 3.8359523330660054e-16 é menor que 0.05, então os dados não estão normalmente distribuidos
```
